```python
airline_flights = df.groupby("Airline")["Flight_ID"].count()

airline_flights.sort_values(ascending=False)
```

```python
airline_analysis = df.groupby("Airline").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Passengers=("Passengers", "mean"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
)

airline_analysis.sort_values("Total_Passengers", ascending=False)
```

```python
origin_analysis = df.groupby("Origin").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum")
)

origin_analysis.sort_values("Total_Passengers", ascending=False)
```

```python
destination_analysis = df.groupby("Destination").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum")
)

destination_analysis.sort_values("Total_Passengers", ascending=False)
```

```python
class_analysis = df.groupby("Travel_Class").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Ticket_Price=("Average_Ticket_Price", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
)

class_analysis
```

```python
weather_analysis = df.groupby("Weather").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
)

weather_analysis.sort_values("Average_Delay", ascending=False)
```

```python
booking_analysis = df.groupby("Booking_Channel").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Ticket_Price=("Average_Ticket_Price", "mean")
)

booking_analysis.sort_values("Total_Passengers", ascending=False)
```

```python
df["Flight_Date"] = pd.to_datetime(df["Flight_Date"])

df["Flight_Date"].head()
```

```python
df["Flight_Month"] = df["Flight_Date"].dt.month

df[["Flight_Date", "Flight_Month"]].head()
```

```python
df["Flight_Day"] = df["Flight_Date"].dt.day_name()

df[["Flight_Date", "Flight_Day"]].head()
```

```python
df["Seat_Utilization"] = (
    df["Passengers"] / df["Seat_Capacity"]
) * 100

df[["Flight_ID", "Passengers", "Seat_Capacity", "Seat_Utilization"]].head()
```

```python
df["Estimated_Revenue"] = (
    df["Passengers"] * df["Average_Ticket_Price"]
)

df[["Flight_ID", "Passengers", "Average_Ticket_Price", "Estimated_Revenue"]].head()
```

```python
df.sort_values(
    "Estimated_Revenue",
    ascending=False
).head(10)
```

```python
route_analysis = df.groupby(
    ["Origin", "Destination"]
).agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Estimated_Revenue=("Estimated_Revenue", "sum")
)

route_analysis.sort_values(
    "Total_Passengers",
    ascending=False
).head(10)
```

```python
delayed_flights = df[df["Flight_Status"] == "Delayed"]

print("Number of Delayed Flights:", len(delayed_flights))
print("Average Delay:", delayed_flights["Delay_Minutes"].mean())
```
